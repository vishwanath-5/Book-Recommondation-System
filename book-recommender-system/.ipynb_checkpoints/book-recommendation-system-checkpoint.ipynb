# Modifing ratings and ages for books

```python
user.head()
```

```python
book_statistics = user.groupby('ISBN').agg(
    AverageRating=('Rating', 'mean'),
    MinAge=('Age', 'min'),
    MaxAge=('Age', 'max')
)
```

```python
book_statistics.head()
```

```python
book_statistics['MinAge'] = book_statistics['MinAge'].replace(0, np.nan)
```

```python
book_statistics.isnull().sum()
```

# Removing data whose average rating is less than 7

```python
book_statistics = book_statistics[book_statistics['AverageRating']>=7]
```

# stemming and Applying Bag of words for books

```python
import nltk
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return ' '.join(y)
```

```python
data['Tag'].apply(stem)
```

```python
data['Tag'] = data['Tag'].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
vectors = cv.fit_transform(data['Tag']).toarray()
```

```python
vectors
```

```python
cv.get_feature_names_out().tolist()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# cosine_similarity(vectors)
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity[0]
```

```python
sorted(list(enumerate(similarity[0])), reverse=True, key = lambda x: x[1])
```

```python
def recommend(book):
    book_index = data[data['Title']==book].index[0]
    distances = similarity[book_index]
    book_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]
    for i in book_list:
        print(data.iloc[i[0]]['Title'])
```

```python
recommend("El Misterio De Salem's Lot/Salem's Lot")
```
